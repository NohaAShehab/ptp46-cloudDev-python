```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self.email = email
        self.salary = salary

    @staticmethod  # this is a helper function, doesn't depend on the object or the class ?
    def cal_net_salary(salary):
        return salary* .8


emp1 = Employee("Ahmed", 'ahmed@gmail.com', 4000)
print(emp1.cal_net_salary(50000))
print(Employee.cal_net_salary(emp1.salary))
```
