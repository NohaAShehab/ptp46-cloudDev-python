## ✨ Try-Except-Else Pattern
> 🎯 **Else block:** Executes ONLY if no exception occurred in try block
> 
> ✅ **Success path:** Print result only when calculation succeeds!

```python
try:
    num = int(input("Please enter a number: "))
    res = 4 / num
except Exception as e:
    print(f"--- problem happened {e}")
else:
    "process is done without exceptions"
    print(f"--- res=  {res}")
finally:
    print("--- this block will be executed always")

print("-------------------- the end -------------------")
```

## 🎯 Alternative: Print Inside Try
> 💡 **Different approach:** Print result inside try block instead of using else
> 
> ⚠️ **Less clean:** Mixes logic with try block

```python
def sum_num() -> int:
    try:
        num1 = int(input("Please enter a number: "))
        num2 = int(input("Please enter a number: "))
    except Exception as e:
        print(f"--- problem happened {e}")
        return False
    else:
        return num1 + num2
    finally:
        """ finally execution precedes the return """
        print("--- Process completed ")
    print("------------- process completed successfully")


print(sum_num())
```

## 🎪 Try-Except-Else-Finally Pattern
> 🌟 **Complete pattern:** Finally block ALWAYS executes (success or failure)
> 
> 🔒 **Cleanup code:** Perfect for closing files, connections, releasing resources!
