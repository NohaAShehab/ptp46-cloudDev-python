## 🔭 Lexical Scope in Python
> 🌐 **Variable scope:** Understanding where variables can be accessed in your code
> 
> 📚 **Types:** Global scope, Local scope, and Nested scope (nonlocal)

```python
name = 'Ahmed' # global variable
```

## 🌍 Global Variable Declaration
> 📍 **Accessible everywhere:** Variables defined at module level are global
> 
> ✅ **Can be read:** From anywhere in the module

```python

def sum_num():
    """
    num1 and num2 and res  are local variables can be accessed inside the function only
    :return:
    """
    num1 = 10
    num2 = 20
    res = num1 + num2
    print(f'num1= {num1}, num2= {num2}, res= {res}')

sum_num()
print("-----------------")
print(num1, num2)
```

## 🔒 Local Variables Example
> ⚠️ **Scope limitation:** Variables inside functions are local - only accessible inside that function
> 
> 💥 **Error:** Trying to access num1, num2 outside the function will fail!

## Accessing global variable from different places

```python
# access inside the script
print(name) #read value
name='Ahmed Mohamed' # update the value
print(name)
```

```python
# read global variable from inside the function

def print_name():
    print(f'name= {name}')
print_name()
```

## ✅ Reading & Updating in Script
> 📖 **Read global variable:** Print its value
> 
> ✏️ **Update global variable:** Reassign new value at module level

```python
# modify the global variable value ?

def modify_name():
    name = "updated" # new local variable
    print(f'name after update = {name}')

modify_name()
print(name)
```

## 📖 Reading Global from Function
> ✅ **Read access works:** Functions can read global variables without any special keyword
> 
> 🎯 **No modification:** Just reading the value

```python

def modify_global_name():
    """ please don't create new local variable , use the global one """
    global name
    name = "updated" # new local variable
    print(f'name after update = {name}')

modify_global_name()
print(name)
```

## ⚠️ Shadowing Global Variable
> 🎭 **Local shadows global:** Assignment creates NEW local variable, doesn't modify global
> 
> 🔴 **Common mistake:** Global variable remains unchanged!

## 🌐 Using global Keyword
> ✅ **Modify global:** Use `global` keyword to modify the actual global variable
> 
> 🎯 **Now it works:** Global variable is truly updated!

```python
def outer_function( ):
    username= 'iti'

    def print_username():
        """ from inner function , we read parent local variables """
        print(f'username= {username}')
    print_username()

outer_function()
```

```python
def outer_function( ):
    username= 'iti'

    def modify_username():
        username= 'Ali'  # create new local variable
        print(f'username= {username}')
    modify_username()
    print(username)

outer_function()
```

## 🔴 Nested Function Not Called
> ⚠️ **Nothing happens:** Inner function is defined but never called
> 
> 💡 **No output:** Function definition doesn't execute the code

```python
def outer_function( ):
    username= 'iti'

    def modify_username():
        """ please don't create new local variable , use the parent's one """
        nonlocal username
        username= 'Ali'  # create new local variable
        print(f'username= {username}')
    modify_username()
    print(username)

outer_function()
```

## ✅ Reading Parent's Local Variable
> 📖 **Inner function reads parent's variable:** Nested functions can access enclosing scope
> 
> 🎯 **Works!** Inner function called and reads username

## 🌍 Creating Global from Function
> ✨ **Global creates variable:** Function creates a new global variable 'message'
> 
> 💡 **Accessible after call:** Variable persists in global scope

```python
def A():
    username='Ahmed'
    def B():
        def C():
            def D():
                def E():
                    global  track
                    track = 'DevOps'
                E()
            D()
        C()
    B()
A()
```

## ⚠️ Nonlocal Without Parent Variable
> 🔴 **Error will occur:** `nonlocal` requires the variable to exist in parent scope
> 
> 💥 **Won't work:** Can't use nonlocal to create new variables!

## 🎢 Deep Nesting - Scope Chain
> 🪆 **Multiple levels:** Inner functions can access variables from any enclosing scope
> 
> 🔍 **Scope chain:** E → D → C → B → A (searches outward until found)
