```python

def sum_num():
    """
    num1 and num2 and res  are local variables can be accessed inside the function only
    :return:
    """
    num1 = 10
    num2 = 20
    res = num1 + num2
    print(f'num1= {num1}, num2= {num2}, res= {res}')

sum_num()
print("-----------------")
print(num1, num2)
```

## Accessing global variable from different places

```python
# access inside the script
print(name) #read value
name='Ahmed Mohamed' # update the value
print(name)
```

```python
# read global variable from inside the function

def print_name():
    print(f'name= {name}')
print_name()
```

```python
# modify the global variable value ?

def modify_name():
    name = "updated" # new local variable
    print(f'name after update = {name}')

modify_name()
print(name)
```

```python

def modify_global_name():
    """ please don't create new local variable , use the global one """
    global name
    name = "updated" # new local variable
    print(f'name after update = {name}')

modify_global_name()
print(name)
```

## Nonlocal

```python
def outer_function( ):
    username= 'iti'

    def print_username():
        print(f'username= {username}')

outer_function()
```

```python
def outer_function( ):
    username= 'iti'

    def print_username():
        """ from inner function , we read parent local variables """
        print(f'username= {username}')
    print_username()

outer_function()
```

```python
def outer_function( ):
    username= 'iti'

    def modify_username():
        username= 'Ali'  # create new local variable
        print(f'username= {username}')
    modify_username()
    print(username)

outer_function()
```

```python
def outer_function( ):
    username= 'iti'

    def modify_username():
        """ please don't create new local variable , use the parent's one """
        nonlocal username
        username= 'Ali'  # create new local variable
        print(f'username= {username}')
    modify_username()
    print(username)

outer_function()
```
