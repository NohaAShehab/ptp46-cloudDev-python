## 🎨 Creating a Set
> 🔮 **Set basics:** Sets automatically remove duplicates and are unordered collections of unique elements

```python
s = set()

mys = {"iti", "iti", "iti", "iti", 324,323.3, True, None,
       "cloud", "database", ("python", "mongo", "Node")}

print(mys)
```

```python
"""
    1- remove duplicates
    2- un-ordered datatype
    3- no index for accessing elements
"""

""" set mutable or immutable """
mys.add('new element')
print(mys)
```

## 📚 Set Properties & Mutability
> ✨ **Key features:** Sets remove duplicates, are unordered, have no indexing, but are mutable (can be modified)

```python
popped_element = mys.pop() #remove element from set
```

```python
print(popped_element)
```

## 🗑️ Removing Elements with Pop
> 🎲 **Random removal:** `pop()` removes and returns an arbitrary element from the set
