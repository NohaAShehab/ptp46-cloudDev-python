## 🔍 Checking Function Type
> 🏷️ **Type verification:** Functions are first-class objects in Python with type 'function'

```python
def get_full_name():
    first_name = input("Enter your first name: ")
    last_name = input("Enter your last name: ")
    fullname = f'{first_name} {last_name}'
    return fullname

username = get_full_name()
print(username)
```

```python
def get_full_name2():
    first_name = input("Enter your first name: ")
    last_name = input("Enter your last name: ")
    fullname = f'{first_name} {last_name}'
    return  first_name, last_name, fullname # tuple

username2 = get_full_name2()
print(username2)
```

## ➕ Function with Required Parameters
> 📌 **Parameters:** Define functions that require specific arguments to be passed

```python

def sum_num2(num1, num2=10):
    print(f"num1={num1}, num2={num2}")
    res = num1 + num2
    print(f'res= {res}')
    return res

sum_num2(3,4)
sum_num2(3)
```

## ⚠️ Syntax Error: Non-default After Default
> 🚫 **Invalid syntax:** Default parameters must come after non-default parameters in function definition

```python

def sum_num3(num1=100, num2):
    print(f"num1={num1}, num2={num2}")
    res = num1 + num2
    print(f'res= {res}')
    return res
```
