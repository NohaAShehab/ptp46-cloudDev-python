<strong style="color:inherit;">While Loop: 1 to 10</strong><br>
<span style="color:inherit;">Build a list of numbers using a counter and while loop.</span>
</div>



```python
# generate list of numbers from 1 to 10
counter = 1
nums = []
while counter <= 10:
    nums.append(counter)
    counter += 1

print(nums)
```

<div style="background:rgba(14,165,233,0.08);border:1px solid rgba(14,165,233,0.35);border-left:4px solid rgba(14,165,233,0.9);padding:10px 12px;border-radius:8px;">
<strong style="color:inherit;">While Loop: Odd Numbers</strong><br>
<span style="color:inherit;">Append numbers by stepping the counter by 2.</span>
</div>



```python
# generate list of numbers from 1 to 10
counter = 1
nums = []
while counter <= 10:
    nums.append(counter)
    counter += 2

print(nums)
```

<div style="background:rgba(14,165,233,0.08);border:1px solid rgba(14,165,233,0.35);border-left:4px solid rgba(14,165,233,0.9);padding:10px 12px;border-radius:8px;">
<strong style="color:inherit;">Range Object</strong><br>
<span style="color:inherit;">Create a range and print its representation.</span>
</div>



```python
myr = range(10)
print(myr) # iterable object
```

<div style="background:rgba(14,165,233,0.08);border:1px solid rgba(14,165,233,0.35);border-left:4px solid rgba(14,165,233,0.9);padding:10px 12px;border-radius:8px;">
<strong style="color:inherit;">Range to List</strong><br>
<span style="color:inherit;">Convert a range into a concrete list.</span>
</div>



```python
print(list(myr))
```

<div style="background:rgba(14,165,233,0.08);border:1px solid rgba(14,165,233,0.35);border-left:4px solid rgba(14,165,233,0.9);padding:10px 12px;border-radius:8px;">
<strong style="color:inherit;">Iterate Over Range</strong><br>
<span style="color:inherit;">Loop through a range and print each value.</span>
</div>



```python
for i in myr:
    print(i)
```
