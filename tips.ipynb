```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name # public
        self.email = email  # protected
        self.salary = salary

    def __str__(self):
        return f"{self.name}"
    def __repr__(self):
        return f"Employee(name={self.name}, email={self.email}, salary={self.salary})"

    def __len__(self):
        # must return with int
        return len(self.__dict__)


emp = Employee("test", "test@gmail.com", 80000)
print(emp.__dict__)
print(emp)
print(emp.__repr__())
emp.city = 'cairo'
print(len(emp))
```
