## 🔐 Password Login Loop (Version 1)
> 🔄 **Basic attempt:** A simple password check with break statement, but account locks even on success!

```python
for  i in range(3):
    password = input("please enter your password: ")
    if password == "abc":
        print("-- logged in successfully --")
        break
    print("------ incorrect password, please try again------ --")

print("---- Account is locked----")
```

## 🔧 Password Loop (Version 2)
> 🎯 **Improved with index check:** Only locks account if all 3 attempts failed (i==2)

```python
logged_in = False
for  i in range(3):
    password = input("please enter your password: ")
    if password == "abc":
        print("-- logged in successfully --")
        logged_in = True
        break
    print("------ incorrect password, please try again------ --")

if not logged_in :
    print("---- Account is locked----")
```

```python
for  i in range(3):
    password = input("please enter your password: ")
    if password == "abc":
        print("-- logged in successfully --")
        break
    print("------ incorrect password, please try again------ --")
else:
    """ this block will be executed only if the loop wasn't broken"""
    print("---- Account is locked----")
```
