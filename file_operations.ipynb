## 🔓 Opening a File - Basic Try-Except
> 🛡️ **Error Handling:** Using try-except to handle file not found errors gracefully
> 
> 📖 **Mode 'r':** Opens file in read mode (file must exist!)

```python
"""
    open(filename, mode)
    mode --> r for read
    w --> for write
    a --> append
"""

try:
    file_object = open("users.txt", "r")
except Exception as e:
    print(e)
else:
    print(file_object)
    file_object.close()
```

## ✅ Successful File Open
> 🎉 **File exists now!** The file object is created successfully

> ⚠️ **Remember:** Always close the file after operations

## 📖 Reading File Content
> 📄 **Read method:** Get all file content as a single string

```python
"""
    open(filename, mode)
    mode --> r for read
    w --> for write
    a --> append
"""

try:
    file_object = open("users.txt", "r")
except Exception as e:
    print(e)
else:
    print(file_object)
    data  = file_object.read()
    print(data, type(data))
    file_object.close()
```

## 📋 Reading Lines as List
> 📚 **readlines() method:** Returns all lines as a list of strings
> 
> 🔍 **Notice:** Each line includes the newline character `\n`

```python
# readlines
"""
    open(filename, mode)
    mode --> r for read
    w --> for write
    a --> append
"""

try:
    file_object = open("users.txt", "r")
except Exception as e:
    print(e)
else:
    print(file_object)
    lines  = file_object.readlines()
    print(lines, type(lines))
    file_object.close()
```

## 🔄 Iterating Through File Lines
> 🎯 **File iteration:** Iterate directly over the file object to process lines one by one
> 
> 💪 **Memory efficient:** Better for large files!

```python
# readlines
"""
    open(filename, mode)
    mode --> r for read
    w --> for write
    a --> append
"""

try:
    file_object = open("users.txt", "r")
except Exception as e:
    print(e)
else:
    lines = []
    for line in file_object:
        lines.append(line)

    print(lines, type(lines))
    file_object.close()
```

## 🌟 Using Context Manager (with statement)
> ✨ **Best practice:** The `with` statement automatically closes the file
> 
> 🎊 **No need to call close()** - Python handles it for you!

```python
try:
    with open("users.txt", "r") as file_obj:
        print(file_obj.read())
except Exception as e:
    print(e )
```

## ⚠️ File Pointer Issue
> 🔴 **Problem:** After reading, the file pointer is at the end
> 
> 📍 **Result:** readlines() returns empty list because there's nothing more to read!

```python
try:
    with open("users.txt", "r") as file_obj:
        data = file_obj.read()
        lines = file_obj.readlines()
        print(data, lines)
except Exception as e:
    print(e )
```

## 🔄 Using seek() to Reset Pointer
> 🎯 **seek(0):** Moves the file pointer back to the beginning
> 
> ✅ **Now both read operations work!**

```python
try:
    with open("users.txt", "r") as file_obj:
        data = file_obj.read()
        file_obj.seek(0)
        lines = file_obj.readlines()
        print(data, lines)
except Exception as e:
    print(e )
```

## 📄 Opening File in Write Mode
> 🆕 **Creating file:** Mode 'w' creates a new file if it doesn't exist
> 
> ⚠️ **Warning:** If file exists, old content will be deleted!

```python
try:
    file_object = open("students.txt", "w")
except Exception as e:
    print(e)
else:
    print(file_object)
    file_object.write("Hello world")
    file_object.close()
```

## ✏️ Writing Content to File
> 📝 **write() method:** Writes a string to the file
> 
> 🎯 **Result:** "Hello world" is now saved in the file

```python
try:
    file_object = open("students.txt", "w")
except Exception as e:
    print(e)
else:
    print(file_object)
    file_object.write("Hello world")
    file_object.writelines(["new", "test", "lines"])
    file_object.close()
```

## 🗑️ Opening Without Writing
> ⚠️ **Content deleted!** Opening in 'w' mode without writing leaves file empty
> 
> 🔴 **Be careful:** This overwrites any existing content

```python
try:
    file_object = open("students.txt", "w")
except Exception as e:
    print(e)
else:
    print(file_object)
    file_object.write("Hello world\n")
    file_object.writelines(["new\n", "test\n", "lines\n"])

    file_object.close()
```
