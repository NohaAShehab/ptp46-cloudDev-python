```python
iti = 'information Technology Institute'

print(iti.count('i'))
```

```python
# get index of char ?
print(iti.index('i'))
```

<div style="background:rgba(14,165,233,0.08);border:1px solid rgba(14,165,233,0.35);border-left:4px solid rgba(14,165,233,0.9);padding:10px 12px;border-radius:8px;">
<strong style="color:inherit;">While Loop Over String</strong><br>
<span style="color:inherit;">Scan the string and print matching indexes.</span>
</div>



```python
# loop over the string ==> iterable ?
counter = 0
while counter <len(iti):
    if iti[counter] == 'i':
        print(counter)
    counter += 1
```

<strong style="color:inherit;">Manual Replace</strong><br>
<span style="color:inherit;">Build a new string by replacing characters.</span>
</div>



```python
# replace i with $

newstring = ''
for char in iti:
    if char=='i':
        newstring += '$'
    else:
        newstring += char

print(newstring)
```
