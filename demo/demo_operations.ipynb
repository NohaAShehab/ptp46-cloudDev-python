```python
try:
    with open("users.txt") as f:
        data = f.readlines()
        users= []
        for line in data:
            line = line.strip()
            user_info = line.split(':')
            users.append(user_info)

        else:
            print("--- All users : [] ")
            print(users)
except Exception as e:
    print(e)
```
