## 📖 Reading Text File with Processing
> 📄 **File parsing:** Read users.txt and split each line by ':' delimiter
> 
> 🔄 **For-else pattern:** Process all lines then print final result!

```python
try:
    with open("users.txt") as f:
        data = f.readlines()
        users= []
        for line in data:
            line = line.strip()
            user_info = line.split(':')
            users.append(user_info)

        else:
            print("--- All users : [] ")
            print(users)
except Exception as e:
    print(e)
```

## ✍️ Appending User to Text File
> 📝 **User input:** Collect user information from console
> 
> ➕ **Append mode:** Add new user to existing file without deleting old data

```python

id = input("Please enter the id of the user: ")
name= input("Please enter the name of the user: ")
email = input("Please enter the email of the user: ")
password = input("Please enter the password of the user: ")
user_info = f"{id}:{name}:{email}:{password}"
try:
    with open("users.txt", 'a') as f:
        f.write(user_info)
except Exception as e:
    print(e)
```

## 📖 Reading JSON File (Raw String)
> 📄 **Basic read:** Get JSON content as a plain string
> 
> ⚠️ **Not parsed yet:** Still a string, not a Python object!

```python
import json
try:
    with open("users.json") as f:
        data = f.read() # string
        # read json content from string
        data = json.loads(data)
        print(data, type(data))

        print(data)
except Exception as e:
    print(e)
```

## 🔄 Parsing JSON String
> 🎯 **json.loads():** Convert JSON string to Python list/dict
> 
> ✨ **Now it's usable:** Can access data as Python objects!

```python
import json

try:
    with open("users.json") as f:
        users = json.load(f)  # accept file object
        print(users, type(users))
except Exception as e:
    print(e)
```

## ❌ Writing Dict Directly (Error)
> 🔴 **Type error:** Cannot write dict directly to file - needs to be string!
> 
> ⚠️ **Common mistake:** Files only accept strings, not Python objects

```python

id = input("Please enter the id of the user: ")
name= input("Please enter the name of the user: ")
email = input("Please enter the email of the user: ")
password = input("Please enter the password of the user: ")
user_info = {
    "id": id,
    "name": name,
    "email": email,
    "password": password
}
try:
    with open("users.json", 'a') as f:
        f.write(str(user_info))
except Exception as e:
    print(e)
```

```python
def read_all_users():
    users = []
    try:
        with open("users.json") as f:
            users = json.load(f)  # accept file object
            print(users, type(users))
    except Exception as e:
        print(e)
    return users
```
