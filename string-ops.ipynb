## 🔄 String Replacement

Here's a manual approach to replace characters using a loop:

```python

# Replace matching characters

```

---

### Method 2: Using `.replace()` Method ⚡

The `.replace()` method is **much simpler and more efficient**:


- ✅ One line of code

> **💡 Tip:** You can limit the number of replacements!


The `rstrip()` method removes whitespace (or specified characters) from the **right end** of a string.

**Related Methods:**
- `rstrip()` - Remove from right
- `lstrip()` - Remove from left  
- `strip()` - Remove from both ends


<div style="background:rgba(14,165,233,0.08);border:1px solid rgba(14,165,233,0.35);border-left:4px solid rgba(14,165,233,0.9);padding:10px 12px;border-radius:8px;">

```python
text = "@     welcome     @"

```

---

> **⚠️ Important:** `rstrip()` only removes characters from the **end**!

```python
text = "  welcome  $  "
text.rstrip("@$")  # Nothing removed (no @ or $ at the end)
```

## 🔍 The `in` Operator

The `in` operator checks if a character or substring exists within a string.

**Returns:** `True` or `False`

**Syntax:**
```python

```

<strong style="color:inherit;">Manual Replacement Loop</strong><br>
<span style="color:inherit;">Build a new string while replacing characters.</span>
</div>



````python
message = "We do love python much."
newstring = ""
# replace o with @

for abbass in message:
    # print(abbass)
    if abbass=='o':
        newstring +='@'
    else:
        newstring += abbass

print(newstring)
````
